```python
import re

#compile once
ip_pattern=re.compile(r"\d{1,3}\.\d{1,3}\.d{1,3}")
#Resuse many time--faster than calling re.findall()repeatedly
log__lines=["request from 192.168.1.10","Blocked 10.0.0.5","ok 172.16.0.1"]
for line in log__lines:
  match= ip_pattern.search(line)
  if match:
    print("IP adderess found:",match.group())
```

```python
import re

pattern = r"^[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}$"

def is_valid_email(email):
    return bool(re.match(pattern, email))

print(is_valid_email("anu.sai@college.edu"))
print(is_valid_email("bad-email@@x"))
print(is_valid_email("no-at-symbol.com"))
```

```python
import re

line="2026-08-10 14:22:07 ERROR Disk usage above 90% on /dev/sdal"
pattern=r"(\d{4}-\d{2}-\d{2}:\d{2})"
match=re.match(pattern,line)
if match:
  print("timestamp:",match.group(1))
  print("Level:",match.group(2))
  print("message:",match.group(3))
```
